```python
import pandas as pd

df = pd.read_excel("sydney_property_sale -cleaned_Data.xlsx")

print("Dataset shape:", df.shape)
display(df.head())
```

```python
# Sale price
price = df["SOLD_PRICE_CLEAN"].dropna()

print("Number of properties with known sale price:", len(price))
print("Mean:", price.mean())
print("Median:", price.median())
print("Minimum:", price.min())
print("Maximum:", price.max())
print("Standard deviation:", price.std())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(price, bins=15, edgecolor="black")
plt.xlabel("Sale Price ($)")
plt.ylabel("Number of Properties")
plt.title("Distribution of Sydney Property Sale Prices")
plt.show()
```

```python
# Compare sale prices by suburb

suburb_stats = df.groupby("SUBURB")["SOLD_PRICE_CLEAN"].agg(
    ["count", "mean", "median", "min", "max"]
)

print(suburb_stats)

# Bar chart of median sale price
plt.figure(figsize=(8,5))

suburb_stats["median"].plot(kind="bar")

plt.xlabel("Suburb")
plt.ylabel("Median Sale Price ($)")
plt.title("Median Property Sale Price by Suburb")
plt.xticks(rotation=0)
plt.show()
```

```python
# Sale price over time

df_time = df.dropna(subset=["SOLD_PRICE_CLEAN"]).copy()
df_time["SALE_DATE"] = pd.to_datetime(df_time["SALE_DATE"], dayfirst=True)

plt.figure(figsize=(10,5))

plt.scatter(
    df_time["SALE_DATE"],
    df_time["SOLD_PRICE_CLEAN"]
)

plt.xlabel("Sale Date")
plt.ylabel("Sale Price ($)")
plt.title("Property Sale Prices Over Time")
plt.xticks(rotation=45)
plt.show()
```

```python
# Check unusually high and low sale prices

price_check = df.dropna(subset=["SOLD_PRICE_CLEAN"]).sort_values(
    "SOLD_PRICE_CLEAN"
)

print("5 lowest sale prices:")
display(price_check[
    ["ID", "SUBURB", "BEDROOMS", "BATHROOMS",
     "PARKING", "PROPERTY_SIZE_M2", "PROPERTY_TYPE",
     "SOLD_PRICE_CLEAN"]
].head(5))

print("\n5 highest sale prices:")
display(price_check[
    ["ID", "SUBURB", "BEDROOMS", "BATHROOMS",
     "PARKING", "PROPERTY_SIZE_M2", "PROPERTY_TYPE",
     "SOLD_PRICE_CLEAN"]
].tail(5))
```

```python
# Check the 10 largest properties

largest = df.dropna(subset=["PROPERTY_SIZE_M2"]).sort_values(
    "PROPERTY_SIZE_M2", ascending=False
)

display(largest[
    ["ID", "SUBURB", "BEDROOMS", "BATHROOMS",
     "PARKING", "PROPERTY_SIZE_M2", "PROPERTY_TYPE",
     "SOLD_PRICE_CLEAN"]
].head(10))
```

```python
# Property size vs sale price

plt.figure(figsize=(8,5))

plt.scatter(
    df["PROPERTY_SIZE_M2"],
    df["SOLD_PRICE_CLEAN"]
)

plt.xlabel("Property Size (m²)")
plt.ylabel("Sale Price ($)")
plt.title("Property Size vs Sale Price")
plt.show()
```

```python
# Average sale price by number of bedrooms

bedroom_prices = df.groupby("BEDROOMS")["SOLD_PRICE_CLEAN"].mean()

plt.figure(figsize=(8,5))

bedroom_prices.plot(kind="bar")

plt.xlabel("Number of Bedrooms")
plt.ylabel("Average Sale Price ($)")
plt.title("Average Sale Price by Number of Bedrooms")
plt.xticks(rotation=0)
plt.show()
```

```python
# Prepare data for modelling

from sklearn.model_selection import train_test_split

model_df = df.dropna(subset=["SOLD_PRICE_CLEAN"]).copy()

# Feature engineering
model_df["LARGE_PROPERTY"] = (
    model_df["PROPERTY_SIZE_M2"] > model_df["PROPERTY_SIZE_M2"].median()
).astype(int)

model_df["PRICE_PER_M2"] = (
    model_df["SOLD_PRICE_CLEAN"] /
    model_df["PROPERTY_SIZE_M2"]
)

model_df["TOTAL_ROOMS"] = (
    model_df["BEDROOMS"] +
    model_df["BATHROOMS"] +
    model_df["PARKING"]
)

X = model_df[
    ["SUBURB", "BEDROOMS", "BATHROOMS", "PARKING",
     "PROPERTY_SIZE_M2", "PROPERTY_TYPE", "SALE_METHOD",
     "LARGE_PROPERTY", "TOTAL_ROOMS"]
]

y = model_df["SOLD_PRICE_CLEAN"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print("Training properties:", len(X_train))
print("Test properties:", len(X_test))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

categorical = ["SUBURB", "PROPERTY_TYPE", "SALE_METHOD"]

numeric = [
    "BEDROOMS", "BATHROOMS", "PARKING",
    "PROPERTY_SIZE_M2", "LARGE_PROPERTY", "TOTAL_ROOMS"
]

preprocessor = ColumnTransformer([
    ("num", SimpleImputer(strategy="median"), numeric),
    ("cat", OneHotEncoder(handle_unknown="ignore"), categorical)
])

models = {
    "Linear Regression": LinearRegression(),
    "Random Forest": RandomForestRegressor(
        n_estimators=200,
        random_state=42
    ),
    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=100,
        random_state=42
    )
}

pipelines = {}

for name, model in models.items():
    pipelines[name] = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])

print("Three models ready.")
```

```python
from sklearn.model_selection import KFold, cross_validate
import numpy as np

kf = KFold(n_splits=5, shuffle=True, random_state=42)

results = []

for name, pipeline in pipelines.items():

    scores = cross_validate(
        pipeline,
        X_train,
        y_train,
        cv=kf,
        scoring={
            "MAE": "neg_mean_absolute_error",
            "RMSE": "neg_root_mean_squared_error",
            "R2": "r2"
        }
    )

    results.append({
        "Model": name,
        "MAE": -scores["test_MAE"].mean(),
        "RMSE": -scores["test_RMSE"].mean(),
        "R2": scores["test_R2"].mean()
    })

results_df = pd.DataFrame(results)

print(results_df)
```

```python
results_df = results_df.sort_values("RMSE")

display(results_df)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

test_results = []

for name, pipeline in pipelines.items():

    pipeline.fit(X_train, y_train)
    predictions = pipeline.predict(X_test)

    test_results.append({
        "Model": name,
        "MAE": mean_absolute_error(y_test, predictions),
        "RMSE": np.sqrt(mean_squared_error(y_test, predictions)),
        "R2": r2_score(y_test, predictions)
    })

test_results_df = pd.DataFrame(test_results)

display(test_results_df)
```

```python
best_model_name = test_results_df.loc[
    test_results_df["RMSE"].idxmin(), "Model"
]

best_model = pipelines[best_model_name]

print("Best model:", best_model_name)
```

```python
best_predictions = best_model.predict(X_test)

plt.figure(figsize=(8,5))

plt.scatter(y_test, best_predictions)

plt.xlabel("Actual Sale Price ($)")
plt.ylabel("Predicted Sale Price ($)")
plt.title(f"Actual vs Predicted Prices - {best_model_name}")

plt.show()
```

```python
# Find the five largest prediction errors

error_df = X_test.copy()

error_df["Actual_Price"] = y_test
error_df["Predicted_Price"] = best_predictions
error_df["Error"] = error_df["Predicted_Price"] - error_df["Actual_Price"]
error_df["Absolute_Error"] = abs(error_df["Error"])

error_df["ID"] = model_df.loc[X_test.index, "ID"]
error_df["ADDRESS"] = model_df.loc[X_test.index, "ADDRESS"]

largest_errors = error_df.sort_values(
    "Absolute_Error",
    ascending=False
).head(5)

display(largest_errors[
    ["ID", "ADDRESS", "SUBURB", "BEDROOMS", "BATHROOMS",
     "PARKING", "PROPERTY_SIZE_M2", "PROPERTY_TYPE",
     "Actual_Price", "Predicted_Price", "Absolute_Error"]
])
```

```python
# Select 10 held-out properties

test_10 = X_test.copy()

test_10["ID"] = model_df.loc[X_test.index, "ID"]
test_10["Actual_Price"] = y_test
test_10["ML_Prediction"] = best_model.predict(X_test)

test_10 = test_10.sample(
    n=10,
    random_state=42
)

display(test_10)
```

```python
comparison = test_10[
    ["ID", "SUBURB", "BEDROOMS", "BATHROOMS",
     "PARKING", "PROPERTY_SIZE_M2",
     "PROPERTY_TYPE", "Actual_Price", "ML_Prediction"]
].copy()

comparison["LLM_Estimate"] = np.nan
comparison["My_Estimate"] = np.nan

display(comparison)
```

```python
comparison["LLM_Estimate"] = [
    1200000,  # PE024
    650000,   # SH033
    3200000,  # SH039
    550000,   # PE021
    2300000,  # SH006
    500000,   # PE023
    600000,   # PA003
    1200000,  # PE040
    1500000,  # PA030
    1200000   # SH019
]
```

```python
comparison["My_Estimate"] = [
    1100000,  # PE024
    700000,   # SH033
    3300000,  # SH039
    550000,   # PE021
    2400000,   # SH006
    500000,   # PE023
    600000,   # PA003
    1200000,  # PE040
    1600000,   # PA030
    1100000    # SH019
]

display(comparison)
```

```python
comparison["ML_Error"] = abs(
    comparison["ML_Prediction"] - comparison["Actual_Price"]
)

comparison["LLM_Error"] = abs(
    comparison["LLM_Estimate"] - comparison["Actual_Price"]
)

comparison["My_Error"] = abs(
    comparison["My_Estimate"] - comparison["Actual_Price"]
)

display(comparison)

print("Average ML error:", comparison["ML_Error"].mean())
print("Average LLM error:", comparison["LLM_Error"].mean())
print("Average human error:", comparison["My_Error"].mean())
```
